### Prediction Before and After Adaptation

#### ULIP, sonn_c, obj_only, rotate_2

```python
with open('../outputs/ulip1_sonn_c_obj_only_rotate_2.log') as fin:
    lines = fin.readlines()
    # first 8 lines are related to Model & Dataset
    logits_lines = lines[8:]
    print(logits_lines[0])    
    assert len(logits_lines) % 3 == 0, f'len(logits_lines) should be integer times of 3, but got {len(logits_lines)}'
```

```python
import ast
import numpy as np
```

```python
# NOTE 
#   1. 找到 hierar cache 预测正确，但 global cache / zero shot 预测错误的
#   2. 找到 hierar cache / global cache 预测正确，但 zero shot 预测错误的

f_obj_id = open('../outputs/expected_ulip1_sonn_c_obj_only_rotate_2.log', 'w')
topk = 5

num_samples = len(logits_lines) // 3 
for i in range(num_samples):
    j = 3*i
    
    hierar_info = logits_lines[j+2]
    hierar_splits = hierar_info.split(',')
    
    hierar_pred = hierar_splits[-3]
    hier_idx = hierar_pred.find(':')
    
    gt = hierar_splits[-2]
    gt_idx = gt.find(':')
    
    # --- 1. hierar cache 预测正确
    if hierar_pred[hier_idx:] == gt[gt_idx:]:
        
        zero_info = logits_lines[j]
        zero_splits = zero_info.split(',')
        
        zero_pred = zero_splits[-3]
        zero_idx = zero_pred.find(':')
        
        # --- 2. zero-shot 预测不正确
        if zero_pred[zero_idx:] != gt[gt_idx:]:
            h_start = hierar_info.find('%:')
            h_end = hierar_info.find(', 	pred_class')
            h_arr = np.array(ast.literal_eval(hierar_info[h_start+3:h_end]))
            h_sum = np.sum(np.exp(h_arr))
            h_idx = np.argpartition(h_arr, -topk)[-topk:]
            h_val = np.exp(h_arr[h_idx]) / h_sum
            f_obj_id.write(f'[{i}]\thierar\t{hierar_pred}\t{gt}\t{h_idx.tolist()}\t{h_val.tolist()}\n')
            
            global_info = logits_lines[j+1]
            global_splits = global_info.split(',')
            global_pred = global_splits[-3]
            
            g_start = global_info.find('):')
            g_end = global_info.find(', 	pred_class')
            g_arr = np.array(ast.literal_eval(global_info[g_start+3:g_end]))
            g_sum = np.sum(np.exp(g_arr))
            g_idx = np.argpartition(g_arr, -topk)[-topk:]
            g_val = np.exp(g_arr[g_idx]) / g_sum
            f_obj_id.write(f'[{i}]\tglobal\t{global_pred}\t{gt}\t{g_idx.tolist()}\t{g_val.tolist()}\n')
            
            z_start = zero_info.find('>:')
            z_end = zero_info.find(', 	pred_class')
            z_arr = np.array(ast.literal_eval(zero_info[z_start+3:z_end]))
            z_sum = np.sum(np.exp(z_arr))
            z_idx = np.argpartition(z_arr, -topk)[-topk:]
            z_val = np.exp(z_arr[z_idx]) / z_sum
            f_obj_id.write(f'[{i}]\tzero\t{zero_pred}\t{gt}\t{z_idx.tolist()}\t{z_val.tolist()}\n')
            
f_obj_id.close()
print('Done!')    
```

#### ULIP-2, SONN_hardest

```python
import ast
import numpy as np
```

```python
with open('../outputs/ulip2_scanobjnn_hardest.log') as fin:
    lines = fin.readlines()
    # first 8 lines are related to Model & Dataset
    logits_lines = lines[8:]
    print(logits_lines[0])    
    assert len(logits_lines) % 3 == 0, f'len(logits_lines) should be integer times of 3, but got {len(logits_lines)}'
```

```python
# NOTE 
#   1. 找到 hierar cache 预测正确，但 global cache / zero shot 预测错误的
#   2. 找到 hierar cache / global cache 预测正确，但 zero shot 预测错误的

f_obj_id = open('../outputs/expected_ulip2_scanobjnn_hardest.log', 'w')
topk = 5

num_samples = len(logits_lines) // 3 
for i in range(num_samples):
    j = 3*i
    
    hierar_info = logits_lines[j+2]
    hierar_splits = hierar_info.split(',')
    
    hierar_pred = hierar_splits[-3]
    hier_idx = hierar_pred.find(':')
    
    gt = hierar_splits[-2]
    gt_idx = gt.find(':')
    
    # --- 1. hierar cache 预测正确
    if hierar_pred[hier_idx:] == gt[gt_idx:]:
        
        zero_info = logits_lines[j]
        zero_splits = zero_info.split(',')
        
        zero_pred = zero_splits[-3]
        zero_idx = zero_pred.find(':')
        
        # --- 2. zero-shot 预测不正确
        if zero_pred[zero_idx:] != gt[gt_idx:]:
            h_start = hierar_info.find('%:')
            h_end = hierar_info.find(', 	pred_class')
            h_arr = np.array(ast.literal_eval(hierar_info[h_start+3:h_end]))
            h_sum = np.sum(np.exp(h_arr))
            h_idx = np.argpartition(h_arr, -topk)[-topk:]
            h_val = np.exp(h_arr[h_idx]) / h_sum
            f_obj_id.write(f'[{i}]\thierar\t{hierar_pred}\t{gt}\t{h_idx.tolist()}\t{h_val.tolist()}\n')
            
            global_info = logits_lines[j+1]
            global_splits = global_info.split(',')
            global_pred = global_splits[-3]
            
            g_start = global_info.find('):')
            g_end = global_info.find(', 	pred_class')
            g_arr = np.array(ast.literal_eval(global_info[g_start+3:g_end]))
            g_sum = np.sum(np.exp(g_arr))
            g_idx = np.argpartition(g_arr, -topk)[-topk:]
            g_val = np.exp(g_arr[g_idx]) / g_sum
            f_obj_id.write(f'[{i}]\tglobal\t{global_pred}\t{gt}\t{g_idx.tolist()}\t{g_val.tolist()}\n')
            
            z_start = zero_info.find('>:')
            z_end = zero_info.find(', 	pred_class')
            z_arr = np.array(ast.literal_eval(zero_info[z_start+3:z_end]))
            z_sum = np.sum(np.exp(z_arr))
            z_idx = np.argpartition(z_arr, -topk)[-topk:]
            z_val = np.exp(z_arr[z_idx]) / z_sum
            f_obj_id.write(f'[{i}]\tzero\t{zero_pred}\t{gt}\t{z_idx.tolist()}\t{z_val.tolist()}\n')
            
            # 现在的问题是这些概率值没做成归一化 + 没取到类名
            
f_obj_id.close()
print('Done!')    
```

#### OpenShape, ModelNet-C

```python
with open('../outputs/openshape_modelnet_c_dropout_local_2.log') as fin:
    lines = fin.readlines()
    # first 8 lines are related to Model & Dataset
    logits_lines = lines[8:]
    print(logits_lines[0])    
    assert len(logits_lines) % 3 == 0, f'len(logits_lines) should be integer times of 3, but got {len(logits_lines)}'
```

```python
# NOTE 
#   1. 找到 hierar cache 预测正确，但 global cache / zero shot 预测错误的
#   2. 找到 hierar cache / global cache 预测正确，但 zero shot 预测错误的

f_os_mn_c = open('../outputs/expected_openshape_modelnet_c_dropout_local_2.log', 'w')
topk = 5

num_samples = len(logits_lines) // 3 
for i in range(num_samples):
    j = 3*i
    
    hierar_info = logits_lines[j+2]
    hierar_splits = hierar_info.split(',')
    
    hierar_pred = hierar_splits[-3]
    hier_idx = hierar_pred.find(':')
    
    gt = hierar_splits[-2]
    gt_idx = gt.find(':')
    
    # --- 1. hierar cache 预测正确
    if hierar_pred[hier_idx:] == gt[gt_idx:]:
        
        zero_info = logits_lines[j]
        zero_splits = zero_info.split(',')
        
        zero_pred = zero_splits[-3]
        zero_idx = zero_pred.find(':')
        
        # --- 2. zero-shot 预测不正确
        if zero_pred[zero_idx:] != gt[gt_idx:]:
            h_start = hierar_info.find('%:')
            h_end = hierar_info.find(', 	pred_class')
            h_arr = np.array(ast.literal_eval(hierar_info[h_start+3:h_end]))
            h_sum = np.sum(np.exp(h_arr))
            h_idx = np.argpartition(h_arr, -topk)[-topk:]
            h_val = np.exp(h_arr[h_idx]) / h_sum
            f_os_mn_c.write(f'[{i}]\thierar\t{hierar_pred}\t{gt}\t{h_idx.tolist()}\t{h_val.tolist()}\n')
            
            global_info = logits_lines[j+1]
            global_splits = global_info.split(',')
            global_pred = global_splits[-3]
            
            g_start = global_info.find('):')
            g_end = global_info.find(', 	pred_class')
            g_arr = np.array(ast.literal_eval(global_info[g_start+3:g_end]))
            g_sum = np.sum(np.exp(g_arr))
            g_idx = np.argpartition(g_arr, -topk)[-topk:]
            g_val = np.exp(g_arr[g_idx]) / g_sum
            f_os_mn_c.write(f'[{i}]\tglobal\t{global_pred}\t{gt}\t{g_idx.tolist()}\t{g_val.tolist()}\n')
            
            z_start = zero_info.find('>:')
            z_end = zero_info.find(', 	pred_class')
            z_arr = np.array(ast.literal_eval(zero_info[z_start+3:z_end]))
            z_sum = np.sum(np.exp(z_arr))
            z_idx = np.argpartition(z_arr, -topk)[-topk:]
            z_val = np.exp(z_arr[z_idx]) / z_sum
            f_os_mn_c.write(f'[{i}]\tzero\t{zero_pred}\t{gt}\t{z_idx.tolist()}\t{z_val.tolist()}\n')
            
            # 现在的问题是这些概率值没做成归一化 + 没取到类名
            
f_os_mn_c.close()
print('Done!')    
```

```python
with open('../outputs/uni3d_omniobject3d_4096pts.log') as fin:
    lines = fin.readlines()
    # first 8 lines are related to Model & Dataset
    logits_lines = lines[8:]
    print(logits_lines[0])    
    assert len(logits_lines) % 3 == 0, f'len(logits_lines) should be integer times of 3, but got {len(logits_lines)}'
```

```python
# NOTE 
#   1. 找到 hierar cache 预测正确，但 global cache / zero shot 预测错误的
#   2. 找到 hierar cache / global cache 预测正确，但 zero shot 预测错误的

f_omni3d_4096 = open('../outputs/expected_uni3d_omni3d_4096pts.log', 'w')
topk = 5

num_samples = len(logits_lines) // 3 
for i in range(num_samples):
    j = 3*i
    
    hierar_info = logits_lines[j+2]
    hierar_splits = hierar_info.split(',')
    
    hierar_pred = hierar_splits[-3]
    hier_idx = hierar_pred.find(':')
    
    gt = hierar_splits[-2]
    gt_idx = gt.find(':')
    
    # --- 1. hierar cache 预测正确
    if hierar_pred[hier_idx:] == gt[gt_idx:]:
        
        zero_info = logits_lines[j]
        zero_splits = zero_info.split(',')
        
        zero_pred = zero_splits[-3]
        zero_idx = zero_pred.find(':')
        
        # --- 2. zero-shot 预测不正确
        if zero_pred[zero_idx:] != gt[gt_idx:]:
            h_start = hierar_info.find('%:')
            h_end = hierar_info.find(', 	pred_class')
            h_arr = np.array(ast.literal_eval(hierar_info[h_start+3:h_end]))
            h_sum = np.sum(np.exp(h_arr))
            h_idx = np.argpartition(h_arr, -topk)[-topk:]
            h_val = np.exp(h_arr[h_idx]) / h_sum
            f_omni3d_4096.write(f'[{i}]\thierar\t{hierar_pred}\t{gt}\t{h_idx.tolist()}\t{h_val.tolist()}\n')
            
            global_info = logits_lines[j+1]
            global_splits = global_info.split(',')
            global_pred = global_splits[-3]
            
            g_start = global_info.find('):')
            g_end = global_info.find(', 	pred_class')
            g_arr = np.array(ast.literal_eval(global_info[g_start+3:g_end]))
            g_sum = np.sum(np.exp(g_arr))
            g_idx = np.argpartition(g_arr, -topk)[-topk:]
            g_val = np.exp(g_arr[g_idx]) / g_sum
            f_omni3d_4096.write(f'[{i}]\tglobal\t{global_pred}\t{gt}\t{g_idx.tolist()}\t{g_val.tolist()}\n')
            
            z_start = zero_info.find('>:')
            z_end = zero_info.find(', 	pred_class')
            z_arr = np.array(ast.literal_eval(zero_info[z_start+3:z_end]))
            z_sum = np.sum(np.exp(z_arr))
            z_idx = np.argpartition(z_arr, -topk)[-topk:]
            z_val = np.exp(z_arr[z_idx]) / z_sum
            f_omni3d_4096.write(f'[{i}]\tzero\t{zero_pred}\t{gt}\t{z_idx.tolist()}\t{z_val.tolist()}\n')
            
            # 现在的问题是这些概率值没做成归一化 + 没取到类名
            
f_omni3d_4096.close()
print('Done!')    
```

```python
import ast
import numpy as np
import matplotlib.pyplot as plt
```

```python
z_acc_li = []
z_ent_li = []
z_ent_total = 0.0

g_acc_li = []
g_ent_li = []
g_ent_total = 0.0

h_acc_li = []
h_ent_li = []
h_ent_total = 0.0

topk = 5

with open('../outputs/ent_acc_openshape_modelnet40.log') as f_ent:
    lines = f_ent.readlines()
    for i, line in enumerate(lines):
        splits = line.split('\t')
        # --- zero
        zero_li = splits[1]
        zero_acc = round(float(splits[2]), 2)
        
        z_arr = np.array(ast.literal_eval(zero_li))
        z_sum = np.sum(np.exp(z_arr))
        z_idx = np.argpartition(z_arr, -topk)[-topk:]
        z_val = np.exp(z_arr[z_idx]) / z_sum
        z_ent_total += np.sum(-z_val * np.log(z_val))
        
        z_acc_li.append(zero_acc)
        z_ent_li.append(z_ent_total/(i+1))
        
        # --- global
        global_li = splits[3]
        global_acc = round(float(splits[4]), 2)
        
        g_arr = np.array(ast.literal_eval(global_li))
        g_sum = np.sum(np.exp(g_arr))
        g_idx = np.argpartition(g_arr, -topk)[-topk:]
        g_val = np.exp(g_arr[g_idx]) / g_sum
        g_ent_total += np.sum(-g_val * np.log(g_val))
        
        g_acc_li.append(global_acc)
        g_ent_li.append(g_ent_total/(i+1))
        
        # --- hierar
        hierar_li = splits[5]
        hierar_acc = round(float(splits[6]), 2)
        
        h_arr = np.array(ast.literal_eval(hierar_li))
        h_sum = np.sum(np.exp(h_arr))
        h_idx = np.argpartition(h_arr, -topk)[-topk:]
        h_val = np.exp(h_arr[h_idx]) / h_sum
        h_ent_total = np.sum(-h_val * np.log(h_val))
        
        h_acc_li.append(hierar_acc)
        h_ent_li.append(h_ent_total/(i+1))
```
